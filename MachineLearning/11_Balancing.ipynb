```python
from pathlib import Path
import numpy as np
import pandas as pd
```

```python
POS_file=Path("/home/main/repositories/RemoteSensing/MachineLearning/dataset/pos_disturbance_one.csv")
NEG_file=Path("/home/main/repositories/RemoteSensing/MachineLearning/dataset/neg_no_disturbances.csv")

random_state = np.random.RandomState(13456)
```

```python
all_poz = pd.read_csv(POS_file, sep=";")
all_neg = pd.read_csv(NEG_file, sep=";")
all_neg["Selected"] = False
```

```python
years = all_poz["Year"].unique()
print(years)
```

```python
year_chunks = []

for y in years:
    poz_y = all_poz[all_poz["Year"] == y]
    print(len(poz_y))

    s_neg = all_neg[all_neg["Selected"] == False].sample(len(poz_y), random_state=random_state)
    all_neg.loc[s_neg.index, "Selected"] = True
    s_neg.drop(columns=["Selected"], inplace=True)

    poz_y["Label"] = "Poz"

    s_neg["Label"] = "Neg"
    s_neg["Year"] = "NoYearForNeg"


    year_chunks.append(pd.concat([poz_y,s_neg], ignore_index=True))


    # for i,r in poz_y.iterrows():
    #     out_all["Label"].append("Poz")
    #     out_all["PosX"].append(r["PosX"])
    #     out_all["PosY"].append(r["PosY"])
    #     out_all["Year"].append(r["Year"])
    # for i,r in s_neg.iterrows():
    #     out_all["Label"].append("Neg")
    #     out_all["PosX"].append(r["PosX"])
    #     out_all["PosY"].append(r["PosY"])
    #     out_all["Year"].append("NoYearForNeg")

```

```python
out_all = pd.concat(year_chunks, ignore_index=True)
```
