```python
import pandas as pd
```

```python
ds = pd.read_csv("/home/main/repositories/RemoteSensing/MachineLearning/dataset/df_balanced.csv",sep=';')
```

```python
years = ds["Year"].unique()

chunks_train=[]
chunks_valid=[]
```

```python
for y in years:
    ds_y = ds[ds["Year"] == y]
    print(f"{y}:")

    ds_y_poz = ds_y[ds_y["Label"]=="Poz"]
    ds_y_neg = ds_y[ds_y["Label"]=="Neg"]
    print(f"    poz: {len(ds_y_poz)}      neg: {len(ds_y_neg)}")


    ds_y_poz_train = ds_y_poz.sample(frac=0.8)
    ds_y_poz_valid = ds_y_poz.copy().drop(ds_y_poz_train.index)
    print(f"    poz_train: {len(ds_y_poz_train)}    poz_valid: {len(ds_y_poz_valid)}")

    ds_y_neg_train = ds_y_neg.sample(frac=0.8)
    ds_y_neg_valid = ds_y_neg.copy().drop(ds_y_neg_train.index)
    print(f"    neg_train: {len(ds_y_neg_train)}    neg_valid: {len(ds_y_neg_valid)}")

    chunks_train.append(pd.concat([ds_y_poz_train, ds_y_neg_train], ignore_index=True))
    chunks_valid.append(pd.concat([ds_y_poz_valid, ds_y_neg_valid], ignore_index=True))

```

```python
out_train = pd.concat(chunks_train, ignore_index=True)
out_valid = pd.concat(chunks_valid, ignore_index=True)

print(f"train: {len(out_train)}    valid: {len(out_valid)}")
```

```python
out_train.to_csv("/home/main/repositories/RemoteSensing/MachineLearning/dataset/train.csv", sep=';', index=False)
out_valid.to_csv("/home/main/repositories/RemoteSensing/MachineLearning/dataset/valid.csv", sep=';', index=False)
```
