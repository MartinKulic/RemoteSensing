```python
import openeo

bbox = {
        "west":  18.51635,
        "south": 48.78376,
        "east":  18.80255,
        "north": 49.04104
        }

connection = openeo.connect(url="openeo.dataspace.copernicus.eu")
connection.authenticate_oidc()
```

```python
time_period = ["2020-06-01", "2020-10-01"]
```

```python
s2cube = connection.load_collection(
    "SENTINEL2_L2A",
    temporal_extent=time_period,
    spatial_extent=bbox,
    bands=["B04", "B08", "CLD"],
   # max_cloud_cover= 50
)
CLDcube = connection.load_collection(
    "SENTINEL2_L2A",
    temporal_extent=time_period,
    spatial_extent=bbox,
    bands=["CLD"],
)
```

```python
#ndvi = s2cube.ndvi(nir = "B08", red = "B04")
```

```python
cloud_probab = CLDcube.band("CLD")
cloud_mask = cloud_probab >= 20
s2_masked = s2cube.mask(cloud_mask)
```

```python
ndvi_masked = s2_masked.ndvi(nir = "B08", red = "B04")
```

```python
job = ndvi_masked.create_job(
    out_format="GTiff",
    title="Monthly maskked NDVI with CLD",
    description="NDVI monthly for 2020-06"
)
job.start_and_wait()
```

```python
results = job.get_results()
results.download_files("~/ndvi_2020-06_CLD_mask")
```

```python
job2 = cloud_probab.create_job(
    out_format="GTiff",
    title="Monthly mask CLD",
    description="Mask for 2020-06"
)
job2.start_and_wait()
```
